```python
import numpy as np
from given_rotations import qr_given_r
from householder_transformation import householder_transformation as ht_qr
import sys

sys.path.append('./matrix_generator')
from Li
```

```python
def pd_matrix(rows, cols):
    matrix = np.random.randint(-10, 10, (rows, cols))
    pd_matrix = (matrix.T @ matrix).astype(float)

    return pd_matrix
```

```python
def bench_compression(matrix):

    tol = 1e-14

    time_results = {}

    start_time = time.time()
    q, r = qr_given_r(matrix)
    end_time = time.time()

    q = np.where(np.abs(q) < tol, 0, q)
    r = np.where(np.abs(r) < tol, 0, r)

    print(f'Given Rotation execution time: {end_time - start_time}')
    qr = q @ r

    equal = np.isclose(qr, matrix).all()
    if equal:
        print(f'Given Rotation corectness: {equal}')

    time_results['given'] = [start_time, end_time, (end_time - start_time)]

    start_time = time.time()
    q, r = ht_qr(matrix)
    end_time = time.time()

    q = np.where(np.abs(q) < tol, 0, q)
    r = np.where(np.abs(r) < tol, 0, r)

    print(f'Householder execution time: {end_time - start_time}')
    qr = q @ r

    equal = np.isclose(qr, matrix).all()
    if equal:
        print(f'Householder transformation corectness: {equal}')

    time_results['householder'] = [start_time, end_time, (end_time - start_time)]

    start_time = time.time()
    q, r = np.linalg.qr(matrix)
    end_time = time.time()

    print(f'Numpy execution time: {end_time - start_time}')
    qr = q @ r
    equal = np.isclose(qr, matrix).all()

    if equal:
        print(f'Numpy corectness: {equal}')

    time_results['numpy'] = [start_time, end_time, (end_time - start_time)]

    ratio1 = time_results['given'][2] / time_results['numpy'][2]
    ratio2 = time_results['householder'][2] / time_results['numpy'][2]
    ratio3 = time_results['given'][2] / time_results['householder'][2]

    print(f'''
        The difference between Given and Numpy is {ratio1} times
        The difference between Householder and Numpy is {ratio2} times
        The difference between Given and Householder is {ratio3} times
    ''')
```

```python
matrix_1 = pd_matrix(5, 5)
bench_compression(matrix_1)
```
